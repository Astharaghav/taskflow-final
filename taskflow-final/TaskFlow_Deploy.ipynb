# 🚀 TaskFlow - Deploy on Google Colab
Run each cell one by one from top to bottom.
**Before running:** Make sure you have set your MONGODB_URI and JWT_SECRET below.

```
# ✅ STEP 1 - Install Node.js on Colab
!apt-get install -y nodejs npm
!node --version
!npm --version
```

```
# ✅ STEP 2 - Install ngrok (to get a public URL)
!pip install pyngrok
from pyngrok import ngrok
```

```
# ✅ STEP 3 - Clone your GitHub repo
# CHANGE THIS to your actual GitHub username and repo name
GITHUB_USERNAME = "YOUR_GITHUB_USERNAME"
REPO_NAME = "taskflow"

!git clone https://github.com/{GITHUB_USERNAME}/{REPO_NAME}.git
%cd {REPO_NAME}
```

```
# ✅ STEP 4 - Set your environment variables
# CHANGE THESE VALUES to your actual MongoDB URI and secret
import os

MONGODB_URI = "mongodb+srv://username:password@cluster0.xxxxx.mongodb.net/taskflow?retryWrites=true&w=majority"
JWT_SECRET  = "taskflowsecretkey123"

env_content = f"""PORT=5000
MONGODB_URI={MONGODB_URI}
JWT_SECRET={JWT_SECRET}
NODE_ENV=production
"""

with open("backend/.env", "w") as f:
    f.write(env_content)

print("✅ .env file created")
```

```
# ✅ STEP 5 - Install backend dependencies
%cd backend
!npm install
%cd ..
```

```
# ✅ STEP 6 - Install and build frontend
%cd frontend
!npm install
!npm run build
%cd ..
```

```
# ✅ STEP 7 - Set your ngrok auth token
# Get FREE token from: https://dashboard.ngrok.com/get-started/your-authtoken
NGROK_TOKEN = "YOUR_NGROK_AUTH_TOKEN_HERE"
ngrok.set_auth_token(NGROK_TOKEN)
```

```
# ✅ STEP 8 - Start the app and get public URL
import subprocess, time

os.environ['MONGODB_URI'] = MONGODB_URI
os.environ['JWT_SECRET']  = JWT_SECRET
os.environ['NODE_ENV']    = 'production'
os.environ['PORT']        = '5000'

# Start Node server in background
proc = subprocess.Popen(
    ["node", "backend/server.js"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT
)

time.sleep(5)  # wait for server to start

# Create public tunnel
public_url = ngrok.connect(5000)
print("\n🚀 YOUR APP IS LIVE AT:")
print(public_url)
print("\n✅ Open this URL in your browser to use TaskFlow!")
print("✅ Share this URL as your Live Demo link")
```

```
# ⏹️ STOP SERVER (run this when you want to stop)
ngrok.kill()
proc.terminate()
print("Server stopped.")
```
